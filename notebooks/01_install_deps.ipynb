01 — Install workbench dependencies
===================================
Goal: install the Python packages needed for S3 access, images, and ONNX export
into THIS Jupyter kernel.

Run this first in the OpenShift AI workbench.
After it finishes: Kernel → Restart Kernel, then continue with 02_*.

Why install into the kernel?
  A terminal `pip` can target another Python. Installing via the notebook
  interpreter avoids ModuleNotFoundError in later cells.

```python
import subprocess
import sys

PACKAGES = ["boto3", "pillow", "matplotlib", "onnx"]

print("Python:", sys.executable)
print("Installing:", ", ".join(PACKAGES))

subprocess.check_call([sys.executable, "-m", "pip", "install", *PACKAGES])

print("\nDone. Restart the kernel (Kernel → Restart), then open 02_check_s3.py")
```
